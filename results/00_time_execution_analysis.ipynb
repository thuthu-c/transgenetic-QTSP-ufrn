## Análise preliminar

````python
df_experiments = pd.read_csv('2023-08-20.csv', delimiter=';')
````

````python
algorithms = df_experiments['algorithm'].unique()
algorithms
````

````python
filenames = df_experiments['filename'].unique()
filenames
````

````python
df_result = []

for filename in filenames:
    for algorithm in algorithms:
        df_file_algo = df_experiments[(df_experiments['filename'] == filename) & (df_experiments['algorithm'] == algorithm)]
        desc = df_file_algo.describe()
        result = desc['cost'].to_dict()
        result['milisec'] = desc['milisec']['mean']
        result['filename'] = filename
        result['algorithm'] = algorithm
        df_result.append(result)

df = pd.DataFrame(df_result)
df_result = pd.DataFrame(df_result)
````

````python
df_result
````

````python
df_result['num_before'] = df_result['filename'].str.extract(r'FIS(\d+)-').astype(int)
df_result['num_after'] = df_result['filename'].str.extract(r'-(\d+)').astype(int)
````

````python
df_result = df_result.sort_values(by=['algorithm', 'num_before', 'num_after']).drop(columns=['num_before', 'num_after'])
````

````python
df_result = df_result.drop(columns=['count', '25%', '50%', '75%'])
````

````python
df_result = df_result.round(2)
````

````python
import pandas as pd

# List of algorithms to compare, excluding CheapestInsertion and NearestNeighborhood
algorithms = [alg for alg in df_result['algorithm'].unique() if alg not in ['CheapestInsertion', 'NearestNeighborhood']]

# Prepare data for each algorithm
dataframes = {}
for algorithm in algorithms:
    df_algorithm = df_result[df_result['algorithm'] == algorithm]
    df_algorithm['filename'] = df_algorithm['filename'].str.replace(r'data/FIS(\d+-\d+)\.txt', r'\1', regex=True)
    df_algorithm = df_algorithm[['filename', 'mean', 'std', 'min', 'max', 'milisec']]
    df_algorithm.columns = ['id', f'media_{algorithm}', f'desvio_{algorithm}', f'min_{algorithm}', f'max_{algorithm}', f'tempo_{algorithm}']
    dataframes[algorithm] = df_algorithm

# Merge the dataframes on the 'id' column
combined_df = pd.DataFrame()
for algorithm, df in dataframes.items():
    if combined_df.empty:
        combined_df = df
    else:
        combined_df = pd.merge(combined_df, df, on='id', how='outer')

# Function to unformat numbers (assuming unformat_abnt is defined)
def unformat_cols(val):
    return unformat_abnt(val) if pd.notnull(val) and isinstance(val, str) else val

# Apply unformatting to numeric columns for comparison
numeric_columns = [col for col in combined_df.columns if col != 'id']
combined_df_unformatted = combined_df[numeric_columns].applymap(unformat_cols)

# Start building the LaTeX table
latex_output = "\\begin{tabular}{|" + "c|" * (1 + 5 * len(algorithms)) + "}\n\\hline\n"
header_row = "ID & " + " & ".join(f"Media {alg} & Desvio {alg} & Min {alg} & Max {alg} & Tempo {alg}" for alg in algorithms) + " \\\\\n\\hline\n"
latex_output += header_row

# Function to format a cell with bold if it's the smallest in the group
def format_min_bold(value, min_value):
    return f"\\textbf{{{value}}}" if value == min_value else value

# Iterate over the DataFrame and construct each row
for index, row in combined_df.iterrows():
    row_values = [row['id']]
    for metric in ['media', 'desvio', 'min', 'max', 'tempo']:
        metric_values = [combined_df_unformatted.at[index, f'{metric}_{alg}'] for alg in algorithms if f'{metric}_{alg}' in row]
        min_value = min(metric_values, default=None)
        formatted_values = [combined_df.at[index, f'{metric}_{alg}'] for alg in algorithms if f'{metric}_{alg}' in row]
        row_values.extend([format_min_bold(formatted, min_value) for formatted, unformatted in zip(formatted_values, metric_values)])
    
    latex_output += " & ".join(row_values) + " \\\\\n\\hline\n"

latex_output += "\\end{tabular}"

print(latex_output)
````

## Melhor solução

````python
def is_smaller_instance(filename):
    # Extrai o número da instância do nome do arquivo
    num = int(filename.split('/')[-1].split('-')[0].replace('FIS', ''))
    return num <= 14
````

````python
def calculate_results(df):
    # Agrupar por filename e execution
    grouped = df.groupby(['filename', 'execution'])
    results = {}

    # Para cada instância, determinar qual algoritmo teve o menor custo
    for name, group in grouped:
        min_cost = group['cost'].min()
        algorithms_with_min_cost = group[group['cost'] == min_cost]['algorithm'].tolist()

        for algo in df['algorithm'].unique():
            if algo not in results:
                results[algo] = {'win': 0, 'tie': 0, 'lose': 0}

            if algo in algorithms_with_min_cost:
                if len(algorithms_with_min_cost) == 1:
                    results[algo]['win'] += 1
                else:
                    results[algo]['tie'] += 1
            else:
                results[algo]['lose'] += 1

    # Calcular as porcentagens e exibir os resultados
    total_instances = df['filename'].nunique() * df['execution'].nunique()

    for algo, counts in results.items():
        win_percentage = (counts['win'] / total_instances) * 100
        tie_percentage = (counts['tie'] / total_instances) * 100
        lose_percentage = (counts['lose'] / total_instances) * 100
        print(f"Algoritmo {algo} foi melhor em {win_percentage:.2f}% das instâncias, empatou em {tie_percentage:.2f}% e perdeu em {lose_percentage:.2f}%")
````

````python
df_experiments = pd.read_csv('2023-08-20.csv', delimiter=';')

# Calcular resultados para instâncias menores
print("Resultados para instâncias menores:")
calculate_results(df_experiments[df_experiments['filename'].apply(is_smaller_instance)])

# Calcular resultados para instâncias maiores
print("\nResultados para instâncias maiores:")
calculate_results(df_experiments[~df_experiments['filename'].apply(is_smaller_instance)])

print("\nResultados para instâncias maiores sem o memetic:")
calculate_results(df_experiments[(~df_experiments['filename'].apply(is_smaller_instance)) & (df_experiments['algorithm'] != 'Memetic')])
````
